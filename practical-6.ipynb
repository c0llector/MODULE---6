```python
def bayes(prior_obstacle, p_reading_if_obstacle, p_reading_if_clear):
    """Posterior probability of an obstacle after one sensor reading."""
    score_obstacle = prior_obstacle * p_reading_if_obstacle
    score_clear = (1.0 - prior_obstacle) * p_reading_if_clear
    denominator = score_obstacle + score_clear
    if denominator == 0:
        raise ValueError("The observation has zero probability.")
    return score_obstacle / denominator
 
 
# Part A: a single observation (no state transition).
print("Single BLOCKED:", round(bayes(0.30, 0.80, 0.20), 4))
print("Single CLEAR:  ", round(bayes(0.30, 0.20, 0.80), 4))
 
# Part C: the HMM model.
observations = ["BLOCKED", "CLEAR", "BLOCKED", "BLOCKED"]
emission = {
    "BLOCKED": (0.80, 0.20),  # P(reading | obstacle), P(reading | clear)
    "CLEAR":   (0.20, 0.80),
}
p_obstacle_after = 0.30  # initial belief at t0
 
print("t  reading  predicted_O  updated_O  updated_C")
for t, reading in enumerate(observations, start=1):
    if t == 1:
        predicted_o = p_obstacle_after  # no transition before t1
    else:
        predicted_o = (
            0.70 * p_obstacle_after
            + 0.20 * (1.0 - p_obstacle_after)
        )
 
    likelihood_o, likelihood_c = emission[reading]
    p_obstacle_after = bayes(predicted_o, likelihood_o, likelihood_c)
    print(
        t, reading,
        f"{predicted_o:.4f}",
        f"{p_obstacle_after:.4f}",
        f"{1.0 - p_obstacle_after:.4f}"
    )
```

```python
print("Prior 0.10:", round(bayes(0.10, 0.80, 0.20), 4))
print("Prior 0.30:", round(bayes(0.30, 0.80, 0.20), 4))
print("Prior 0.50:", round(bayes(0.50, 0.80, 0.20), 4))
print("Prior 0.70:", round(bayes(0.70, 0.80, 0.20), 4))
```
